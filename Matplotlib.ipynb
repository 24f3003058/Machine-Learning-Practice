```python
from matplotlib import pyplot as plt

X=[2,4,6]
Y=[1,3,9]

plt.plot(X,Y,'ko--')
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[0,1,2,3,4,5]
y1=[0, 1, 4, 9, 16, 25]
y2=[0,1,8,27,64,125]

plt.plot(x,y1,marker='o',linestyle='--',color='b',label='y=x^2')
plt.plot(x,y2,marker='o', linestyle='--',color='k',label='y= x^3')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Line Plot')

plt.legend()
plt.grid(True,c='k')
plt.show()
```

```python
import numpy as np
x=np.random.randint(100,size=100)
y=np.random.randint(100,size=100)
plt.scatter(x,y,marker='o',s=30)

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title("Scatter Plot")
plt.show()
```

```python
x=np.random.normal(20,5,1000)
plt.hist(x,bins=13)
plt.title('Histogram' ,fontsize=25)
plt.show()
```

# Bar Chart

```python
import matplotlib.pyplot as plt
categories= ['A','B','C','D']
values=np.random.randint(100,size=4)

plt.bar(categories ,values,color='skyblue' ,width=0.5)

plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Chart')

plt.show()
```

# Pie Chart

```python
import matplotlib.pyplot as plt
categories=['A','B','C']
values=np.random.randint(100,size=3)

plt.pie(values,labels=categories, explode=[0,0.1,0],autopct="%.1f%%",startangle=90,shadow=True)
plt.title("Pie Chart")
plt.show()
```
