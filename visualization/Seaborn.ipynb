# Seaborn Tutorial


```python
import pandas as pd
import numpy as np
import seaborn as sns
from copy import deepcopy
```

```python
data=pd.read_csv('Model_Building_V1.csv')
```

```python
data
```

```python
exploration_set=deepcopy(data[['Age','Experience','Projects','HoursPerWeek',
'EducationYears','Salary']])

corr_matrix=exploration_set.corr(method='pearson')
sns.heatmap(corr_matrix,annot=True)
```

```python
sns.scatterplot(x=data['Age'],y=data['Salary'],hue=data['Projects'])
```

```python
data.hist(bins=50,figsize=(20,20))
```

```python
sns.pairplot(data, diag_kind="hist")
```
