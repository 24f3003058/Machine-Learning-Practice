# Concat in pandas

```python
import pandas as pd
```

```python
temp_data=pd.DataFrame({
    'Date': ['12-02-2023', '13-02-2023', '14-02-2023', '15-02-2023', '16-02-2023'],
    'TempMax': [24.3, 26.9, 23.4, 15.5, 16.1 ]
})

rainfall_data=pd.DataFrame({
    'Date': ['12-02-2023', '13-02-2023', '14-02-2023', '15-02-2023', '16-02-2023'],
    'Rainfall': [0, 3.6, 3.6, 39.8, 2.8 ]})

```

```python
result_col=pd.concat([temp_data,rainfall_data])
print("\nConcatenate along rows:")
print(result_col)
```

```python
result_col=pd.concat([temp_data,rainfall_data],axis=1)
print("\nConcatenate along columns:")
print(result_col)
```

# ignoring index

```python
result_col_false_ignore=pd.concat([temp_data,rainfall_data],axis=0,ignore_index=False)
print("\n Concatenate along columns with ignore_index = False:")
print(result_col_false_ignore)

result_col_true_ignore=pd.concat([temp_data,rainfall_data],axis=0,ignore_index=True)
print("\n Concatenate along columns with ignore_index=True")
print(result_col_true_ignore)
```

# join

```python
result_col_inner=pd.concat([temp_data,rainfall_data],axis=0,join='inner')
print("\n Concatenate with inner join:")
print(result_col_inner)

result_col_outer=pd.concat([temp_data,rainfall_data],axis=0,join='outer')
print('\n Concatenate with outer join:')
print(result_col_outer)
```

```python
data1={
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'Score': [85, 90, 88] 
}

df1= pd.DataFrame(data1)

data2={
    'Name': ['David', 'Eve', 'Charlie'],
    'Age': [27, 32, 35],
    'Score': [82, 88, 88],
    "extra":[100,100,100]
}
df2=pd.DataFrame(data2)

print("DataFrame 1:\n", df1)
print("\nDataFrame 2:\n",df2)
```

```python
result_col_inner2=pd.concat([df1,df2],axis=0,join='inner',ignore_index=True)
result_col_inner2
```

```python
result_col_inner2_1=pd.concat([df1,df2],axis=1,ignore_index=True)
result_col_inner2_1
```

# keys

```python
result_col_key1=pd.concat([df1,df2],axis=0 ,keys=('A','B'))
print("\n Concatenate along columns with keys, axis=0:")
print(result_col_key1)
```

```python
result_col_key2=pd.concat([df1,df2],axis=1,keys=('A','B'))
print("\n Concatenate along columns with keys, axis=1: ")
print(result_col_key2)
```
