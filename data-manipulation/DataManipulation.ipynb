```python
import pandas as pd
import numpy as np
```

```python
customers = pd.DataFrame({
    'Customer_ID': range(1000, 1020),
    'Age': np.random.randint(18, 70, 20),
    'Gender': np.random.choice(['M', 'F'], 20),
    'Income': np.random.choice(['Low', 'Medium', 'High'], 20, p=[0.4, 0.5, 0.1]),
    'Join_Date': pd.to_datetime(np.random.choice(pd.date_range('2020-01-01', '2023-01-01'), 20))
})

transactions = pd.DataFrame({
    'Transaction_ID': range(5000, 5100),
    'Customer_ID': np.random.choice(customers['Customer_ID'], 100),
    'Date': pd.to_datetime(np.random.choice(pd.date_range('2023-01-01', '2023-03-31'), 100)),
    'Amount': np.random.uniform(5, 500, 100).round(2),
    'Category': np.random.choice(['Electronics', 'Clothing', 'Groceries', 'Entertainment'], 100)
})
```

```python
customers
```

```python
transactions.shape
```

Data Merging:

Merge customer data with transaction data
Find customers with no transactions (left join)
Customer Segmentation:

Group customers by age brackets (18-25, 26-35, etc.) and analyze spending
Calculate average transaction amount by income level

```python
# customers with their transactions
merged=pd.merge(customers,transactions,on='Customer_ID',how="inner")
print("Merged (inner) shape:",merged.shape)

#left join
left_merged=pd.merge(customers,transactions,on="Customer_ID",how="left")
#left join keeps all customers along with customers with no transactions
no_transactions=left_merged[left_merged["Transaction_ID"].isna()]['Customer_ID'].unique()
print("Customer with no transactions:",no_transactions.tolist())
```

```python

# Age Brackets

bins=[18,25,35,45,55,65]
labels=['18-25','26-35','36-45','46-55','56-65']
customers['Age_Bracket']=pd.cut(customers['Age'],bins=bins,labels=labels,include_lowest=True)
age_spending=pd.merge(transactions,customers[["Customer_ID","Age_Bracket"]],on="Customer_ID")
age_bracket_summary=age_spending.groupby('Age_Bracket',observed=True)['Amount'].agg(
    Total_Spending='sum',
    Avg_Transactions='mean',
    Num_Transactions='count'
).round(2)

print(age_bracket_summary)
```
