```python
def category(price):
    if price<10:
        return "Category 1"
    elif 10<=price<20:
        return "Category 2"
    elif 20<=price<30:
        return "Category 3"
    elif 30<=price<40:
        return "Category 4"
    else:
        return "Category 5"
```

```python
df["Category"]= df["PRICE"].apply(category)
```
