```python
# DATA REPLACEMENT
df["RM"]=df["RM"].replace(-1,np.nan)
df["RIVERSIDE"]=df["RIVERSIDE"].replace("UNKNOWN",np.nan)
df["AGE"]=df["AGE"].replace(-2,np.nan)
```

```python
df=df.drop(columns=["Category"])
```

```python
X=df.drop(columns=["PRICE"])
```

```python
y=df["PRICE"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=0)
```

```python
from sklearn.preprocessing import OneHotEncoder,MinMaxScaler,StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
mm_column=["CRIM","ZN","POLINDEX","DIS","HIGHWAYCOUNT","TAX","PTRATIO","IMM","BPL"]
```

```python
rm_pipeline=Pipeline(
    [
        ("imputer",SimpleImputer(strategy="median")),
        ("mm_scaler",MinMaxScaler())

    ]
)
rm_pipeline
```

```python
age_pipeline=Pipeline(
    [
        ("impute",SimpleImputer(strategy="mean")),
        ("mm_scaler",MinMaxScaler())
    ]
)
age_pipeline
```

```python
riverside_pipeline=Pipeline(
    [
        ("river_pipeline",SimpleImputer(strategy="most_frequent")),
        ("ohe",OneHotEncoder(sparse_output=False))
    ]
)
riverside_pipeline
```

```python
ct =ColumnTransformer(
    [
      ("mm_scaling",MinMaxScaler(),mm_column),
      ("std_scaling",StandardScaler(),["INDUS"]),
      ("rm_pipe",rm_pipeline,["RM"]),
      ("age_pipe",age_pipeline,["AGE"]),
      ("river_pp",riverside_pipeline,["RIVERSIDE"])

    ],
    remainder="passthrough"
).set_output(transform="pandas")
ct
```

```python
X_train_tf=ct.fit_transform(X_train)
X_test_tf= ct.transform(X_test)
```
